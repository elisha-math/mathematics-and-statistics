# Descriptive Statistics Analysis of Student Marks

## Introduction

This project demonstrates how descriptive statistics can be used to summarise and understand a dataset of student marks.

The analysis uses Python, Pandas, NumPy, and Matplotlib to calculate statistical measures and create visualisations.

The dataset contains 10 illustrative student marks. The project covers measures of central tendency, measures of dispersion, data visualisation, and interpretation of results.

## Objectives

- Calculate the mean, median, and mode.
- Determine the range, population variance, and standard deviation.
- Present the results using tables and charts.
- Interpret the findings and identify limitations.

```python

# Descriptive Statistics Analysis of Student Marks

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Descriptive Statistics Analysis")
print("Libraries imported successfully.")
```

```python
# Load the student marks dataset

from google.colab import files
uploaded = files.upload()

# Read the CSV file
df = pd.read_csv("student_marks.csv")

# Display the dataset
print("Student Marks Dataset")
display(df)
```

```python
# Calculate descriptive statistics

number_of_students = df["Marks"].count()
total_marks = df["Marks"].sum()
mean_mark = df["Marks"].mean()
median_mark = df["Marks"].median()
mode_mark = df["Marks"].mode().tolist()

lowest_mark = df["Marks"].min()
highest_mark = df["Marks"].max()
mark_range = highest_mark - lowest_mark

population_variance = np.var(df["Marks"], ddof=0)
population_std_dev = np.std(df["Marks"], ddof=0)

# Display the results

print("DESCRIPTIVE STATISTICAL RESULTS")
print("-" * 40)
print(f"Number of students: {number_of_students}")
print(f"Total marks: {total_marks}")
print(f"Mean: {mean_mark:.2f}")
print(f"Median: {median_mark:.2f}")
print(f"Mode: {mode_mark}")
print(f"Lowest mark: {lowest_mark}")
print(f"Highest mark: {highest_mark}")
print(f"Range: {mark_range}")
print(f"Population variance: {population_variance:.2f}")
print(f"Population standard deviation: {population_std_dev:.2f}")
```

```python
# Create a summary table of descriptive statistics

summary = pd.DataFrame({
    "Statistical Measure": [
        "Number of Students",
        "Total Marks",
        "Mean",
        "Median",
        "Mode",
        "Lowest Mark",
        "Highest Mark",
        "Range",
        "Population Variance",
        "Population Standard Deviation"
    ],
    "Value": [
        number_of_students,
        total_marks,
        mean_mark,
        median_mark,
        mode_mark[0],
        lowest_mark,
        highest_mark,
        mark_range,
        population_variance,
        population_std_dev
    ]
})

display(summary)
```

```python
# Format the summary table for better presentation

summary["Value"] = summary["Value"].apply(
    lambda x: f"{x:.2f}" if isinstance(x, (int, float)) else x
)

display(summary)
```

```python
# Visualise the distribution of student marks

plt.figure(figsize=(9, 5))

plt.hist(
    df["Marks"],
    bins=6,
    edgecolor="black"
)

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.xticks(range(40, 101, 10))
plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
# Create a box plot of student marks

plt.figure(figsize=(8, 4))

plt.boxplot(
    df["Marks"],
    patch_artist=True,
    labels=["Student Marks"]
)

plt.title("Box Plot of Student Marks")
plt.ylabel("Marks")
plt.grid(axis="y", alpha=0.3)

plt.show()
```

## Interpretation of Results

The analysis was conducted using a dataset of 10 student marks.

- **Average performance:** The mean mark was 65, while the median was 62.5. This suggests that the overall average performance was around 65 marks.
- **Most frequent mark:** The mode was 60, meaning it was the most frequently recorded mark.
- **Variation in marks:** The range was 45, showing a difference of 45 marks between the highest and lowest scores.
- **Spread of marks:** The population standard deviation was approximately 13.23, indicating how far the marks generally spread around the mean.
- **Distribution:** The histogram shows how the marks are spread across score intervals, while the box plot provides a visual summary of the centre and spread of the data.

## Conclusion

The analysis shows that the average mark was 65, although individual student marks varied. Descriptive statistics and visualisations provide a useful way of summarising and understanding a dataset.

## Limitations of the Analysis

1. **Small dataset:** The analysis uses only 10 observations, which is not enough to draw conclusions about a larger student population.
2. **Illustrative data:** The marks are used for demonstration and do not represent actual examination results.
3. **Limited variables:** The dataset contains only student marks. Other factors that might influence academic performance are not considered.

These limitations should be considered when interpreting the results.
